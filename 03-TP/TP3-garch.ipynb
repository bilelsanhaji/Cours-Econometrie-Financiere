# Avant de commencer

**Langage.** `R` ou `Python`, au choix. Les fonctions équivalentes sont dans
`03-TP/correspondance-R-python.md`. Les corrigés existent dans les deux
langages, déposés après la séance.

**Package clé.** `R` : `rugarch`, pour la spécification et l'estimation des
modèles GARCH (`ugarchspec`, `ugarchfit`, `ugarchforecast`). `Python` :
`arch` (`arch_model`), qui spécifie et estime en un seul appel.

**Données.** Indice CAC40 (`^FCHI`), rendements journaliers en log, en %,
sur une période récente (2018–2023 par exemple). Vous pouvez utiliser
n'importe quel autre actif — c'est même recommandé une fois l'exercice 1
terminé, pour préparer le projet.

# Exercice 1 — Les données parlent-elles déjà d'elles-mêmes ?

(a) Téléchargez le CAC40 (`quantmod::getSymbols`) et calculez les
    rendements logarithmiques journaliers, en pourcentage.
(b) Tracez le prix, puis le rendement, l'un sous l'autre. Que remarquez-vous
    sur l'amplitude des variations au fil du temps ?
(c) Calculez moyenne, écart-type, asymétrie, kurtosis, minimum et maximum
    des rendements. Testez la normalité (Jarque-Bera). Concluez.
(d) Tracez l'ACF et la PACF des rendements, puis des rendements **au
    carré**, sur 20 retards. Qu'observez-vous, et pourquoi n'est-ce pas
    surprenant après le chapitre 3 ?

# Exercice 2 — Tester l'effet ARCH

(a) Appliquez un test de Ljung-Box sur les rendements au carré (12
    retards). Concluez.
(b) Appliquez le test ARCH d'Engle (`FinTS::ArchTest`, 12 retards).
    Retrouvez-vous la même conclusion qu'à la question précédente ? Est-ce
    surprenant, au vu du chapitre 3 ?

# Exercice 3 — Spécifier et estimer

Avec `rugarch`, estimez successivement, sur les mêmes données :

(a) un ARCH(1) (`garchOrder = c(1,0)`),
(b) un GARCH(1,1),
(c) un GJR-GARCH(1,1),
(d) un EGARCH(1,1).

Pour chacun, notez $\widehat\alpha_1$ (ou son équivalent) et
$\widehat\beta_1$. Calculez la persistance $\widehat\alpha_1+\widehat\beta_1$
du GARCH(1,1) : est-elle compatible avec ce que vous avez observé à
l'exercice 1(b) ?

# Exercice 4 — Choisir un modèle

(a) Construisez un tableau comparatif AIC/BIC/log-vraisemblance des
    quatre modèles de l'exercice 3. Lequel retenez-vous, et sur quel
    critère ?
(b) Pour le modèle retenu, extrayez les résidus standardisés et
    vérifiez : ACF des résidus, ACF des résidus au carré, QQ-plot,
    histogramme. Le modèle est-il correctement spécifié ?
(c) Pour les modèles GJR et EGARCH, le coefficient d'asymétrie
    ($\varphi_1$ ou $\gamma$ selon le modèle) est-il significatif ? Le
    signe est-il celui attendu pour un effet de levier ?

# Exercice 5 — Visualiser et prévoir la volatilité

(a) Superposez, sur un même graphique, la volatilité conditionnelle
    estimée par les quatre modèles. Zoomez sur une période de crise
    (2020, par exemple) : les modèles divergent-ils davantage à ce
    moment-là ?
(b) Produisez une prévision de volatilité à 20 jours avec le modèle
    retenu à l'exercice 4. Commentez sa forme (converge-t-elle ? vers
    quoi, et à quelle vitesse — cf. exercice 3 de la fiche TD3 sur la
    demi-vie d'un choc).

# Exercice 6 — Value-at-Risk et backtesting

(a) Calculez la VaR à 1 jour, au seuil 95 %, à partir du modèle GARCH(1,1)
    sous hypothèse de normalité :
    $\text{VaR}_{t+1} = \widehat\mu_{t+1} + \widehat\sigma_{t+1}\times q_{5\%}$.
(b) Comptez le nombre de violations (rendement inférieur à la VaR) sur
    tout l'échantillon. Le taux observé est-il proche du taux théorique
    (5 %) ?
(c) Testez formellement la calibration par un test de Kupiec. Concluez.
(d) Reprenez (a)-(c) avec une distribution de Student au lieu de la
    normale (`distribution.model = "std"` dans `ugarchspec`). Le nombre
    de violations change-t-il dans le sens attendu ? Pourquoi ?
