# Avant de commencer

**Langage.** `R` ou `Python`, au choix. Les fonctions équivalentes sont dans
`03-TP/correspondance-R-python.md`. Les corrigés existent dans les deux
langages, déposés après la séance.

**Packages clés.** `R` : `forecast` (`auto.arima`, `checkresiduals`),
`tseries` (`adf.test`, `kpss.test`, `Box.test`, `jarque.bera.test`),
`lmtest` (`bptest`, `coeftest`), `FinTS` (`ArchTest`). `Python` :
`statsmodels` (`ARIMA`, `acorr_ljungbox`, `jarque_bera`, `het_arch`,
`het_white`). Pas d'équivalent direct fiable à `auto.arima` : le corrigé
reconstruit la sélection par AIC à la main (voir
`03-TP/correspondance-R-python.md`).

**Données.** Rendements journaliers d'une action ou d'un indice de votre
choix (par exemple AAPL, 2018–2020, autour de la crise Covid).

# Exercice 1 — Stationnarité : prix vs rendements

(a) Téléchargez les prix de clôture ajustés, calculez les log-rendements
    en %.
(b) Testez la stationnarité des prix, puis des rendements, avec le test
    KPSS (`tseries::ur.kpss` ou `kpss.test`, $H_0$ : stationnarité).
    Les deux tests concluent-ils comme attendu ?
(c) Refaites le test ADF (`adf.test`, $H_0$ : racine unitaire) sur les
    rendements. Les deux tests (KPSS, ADF) sont-ils cohérents entre eux
    (attention à l'hypothèse nulle, qui est inversée d'un test à
    l'autre) ?

# Exercice 3 — Estimation

(a) Estimez un AR(1) (`arima(..., order = c(1,0,0))`) et examinez la
    significativité des coefficients (`lmtest::coeftest`).
(b) Estimez automatiquement un ARMA avec `forecast::auto.arima`.
    L'ordre retenu correspond-il à votre lecture de l'exercice 2 ?
(c) Comparez l'AIC des deux modèles. Lequel retenez-vous sur ce
    critère ?

# Exercice 4 — Diagnostic du modèle retenu

(a) Extrayez les résidus du modèle retenu à l'exercice 3.
(b) Testez la normalité (QQ-plot, `tseries::jarque.bera.test`).
(c) Testez l'absence d'autocorrélation résiduelle (`Box.test(...,
    type = "Ljung-Box")`, à plusieurs ordres : 1, 6, 12).
(d) Testez l'hétéroscédasticité (test de White, à construire "à la
    main" par régression auxiliaire comme au TD2, ou `lmtest::bptest`).
(e) Le modèle est-il congruent (aucun test rejeté) ? Sinon, lequel des
    diagnostics pose problème ?
